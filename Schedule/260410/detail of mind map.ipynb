# Project Title: Federated Learning and Deep Reinforcement Learning for Diabetes Risk Prediction (FedDRL)

---

## I. Achievements

### 1. Data Side: NHANES Data Instantiation
* **Multi-year Data Loading**: Successfully loaded datasets across multiple cycles, including `DEMO_L`, `GLU_L`, `GHB_L`, `BIOPRO_L`, `CBC_L`, and `DIQ_L`.
* **Standardized Cleaning Pipeline**: Established a robust data cleaning workflow within the `DatasetMerge.ipynb` logic.
* **Labeling and Feature Engineering**: Selected 15 key physiological indicators and implemented data labeling based on the `DIQ010` variable.
* **Outcome**: Produced a structured **CLEAN** dataset specifically formatted for training and testing.

### 2. Preliminary Model Simulation: (Tested with Diabetes Dataset)
* **Architecture**: Implemented a deep neural network from scratch using **NumPy** to ensure a fundamental understanding of the optimization process.
* **Algorithm Implementation**: Applied a standard deep learning pipeline encompassing initialization (He/Xavier), normalization, training, validation, and testing.
* **Outcome**: The model achieved convergence within 200 epochs, with the loss decreasing steadily from approximately 0.27 to 0.21. 
* **Performance**: Reached a final test accuracy of **90.56%** using the 2017-2018 NHANES data as a benchmark.
* **Visualization**: Generated training and validation curves using Matplotlib to verify model behavior.

## III. Planned Work

### 1. Federated Learning (FL) Simulation
* **Core Task**: Manually partition the 3,349 cleaned data samples into distinct virtual client nodes to simulate a decentralized environment.
* **Challenge**: Accurately simulating **Non-IID** (Non-Independent and Identically Distributed) data distributions, specifically focusing on Cluster Skew and Quantity Skew.

### 2. Deep Reinforcement Learning (DRL) Agent Implementation
* **Core Task**: Defining the DRL Agent's **State** (e.g., local loss, data size), **Action** (aggregation weights $\alpha$), and **Reward** function.
* **Challenge**: Training the agent to dynamically adjust Federated Learning aggregation weights automatically, replacing the traditional, static **FedAvg** algorithm.
