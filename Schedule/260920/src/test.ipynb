```python
year_groups = {
    "Group1",
    "Group2",
    "Group3",
    "Group4",
    "Group5"
}

for group_name in year_groups:
    if group_name == "Group1":
        X_client, Y_client, global_scaler = pre.load_and_preprocess(f'./data_set/balance_data_{group_name}.csv', fit=True)
        X_train_scaled, X_test_scaled, Y_train, Y_test = pre.prepare_data_splits(X_client, Y_client, test_size=0.2, random_state=42)
        
        X_train_scaled.to_parquet(f"../data/processed/partition_{group_name}_train.parquet")
        X_test_scaled.to_parquet(f"../data/processed/partition_{group_name}_test.parquet")
        Y_train.to_parquet(f"../data/processed/partition_{group_name}_train_labels.parquet")
        Y_test.to_parquet(f"../data/processed/partition_{group_name}_test_labels.parquet")
                          
    else:    
        X_client, Y_client, global_scaler = pre.load_and_preprocess(f'./data_set/balance_data_{group_name}.csv', scaler=global_scaler, fit=False)
        X_train_scaled, X_test_scaled, Y_train, Y_test = pre.prepare_data_splits(X_client, Y_client, test_size=0.2, random_state=42)
        
```
