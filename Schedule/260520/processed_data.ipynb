```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import os
import csv
```

```python
file_years = [
    '21_23', '19_20', '17_18', '15_16', '13_14', '11_12', '09_10', '07_08', '05_06', '03_04', '01_02', '99_00'
]

data = [pd.read_csv(f'./data_set/Nhanes_cleaned_{year}.csv') for year in file_years]

if not os.path.exists("./data_set"):
    os.makedirs("./data_set")
```

```python
temp_list = []

for i in range(len(data)):
    risk_group = data[i].query("DIQ010 == 1.0 or DIQ010 == 3.0").copy()
    temp_list.append(risk_group)
    
all_risk_groups = pd.concat(temp_list, axis=0)
all_risk_groups = all_risk_groups.reset_index(drop=True)

# 存到指定路徑
all_risk_groups.to_csv('./data_set/all_risk_groups.csv', index=False)
```

```python
temp_list = []

for i in range(len(data)):
    health_group = data[i].query("DIQ010 == 2.0").copy()
    temp_list.append(health_group)
    
all_health_groups = pd.concat(temp_list, axis=0)
all_health_groups = all_health_groups.reset_index(drop=True)

# 存到指定路徑
all_health_groups.to_csv('./data_set/all_health_groups.csv', index=False)
```

```python
print('危險的人數:' , len(all_risk_groups))
print('健康的人數:' , len(all_health_groups))
```

```python
temp_list = []

temp_list = all_health_groups[0:len(all_risk_groups)].copy()

balance_data = pd.concat([temp_list, all_risk_groups], axis=0)
balance_data = balance_data.reset_index(drop=True)
balance_data = balance_data.sample(frac=1).reset_index(drop=True)
balance_data.to_csv('./data_set/balance_data.csv', index=False)

print('平衡後的資料集:')
print('危險的人數:' , len(balance_data.query("DIQ010 == 1.0 or DIQ010 == 3.0")))
print('健康的人數:' , len(balance_data.query("DIQ010 == 2.0")))

```
